```python
import pandas as pd
```

```python
df = pd.read_csv('prepd_dataset.csv')
```

```python
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
NUMERIC_FEATURES = nums = df.select_dtypes(exclude=["object"]).columns
```

```python
NUMERIC_FEATURES = ['duration', 'src_bytes', 'dst_bytes', 'wrong_fragment',
       'urgent', 'hot', 'num_failed_logins', 'num_compromised',
        'num_root', 'num_file_creations',
       'num_shells', 'num_access_files', 'num_outbound_cmds',
       'count', 'srv_count', 'serror_rate', 'diff_srv_rate',
       'srv_diff_host_rate', 'dst_host_srv_count', 'dst_host_diff_srv_rate',
       'dst_host_same_src_port_rate', 'dst_host_srv_diff_host_rate',
       'dst_host_srv_rerror_rate']
```

```python
CAT_FEATURES = nums = df.select_dtypes(exclude=["number"]).columns
```

```python

CAT_FEATURES = [
    'logged_in',
    'protocol_type',
    'service',
    'flag',
    'root_shell',
    'su_attempted',
    'is_host_login',
    'is_guest_login'
]
```

```python
# for col in CAT_FEATURES:
#     if x_train[col].apply(lambda x: isinstance(x, (int))).any():
#         print(f"⚠️ Column '{col}' contains non-scalar values!")
```

```python
len(NUMERIC_FEATURES) + len(CAT_FEATURES)
```

```python
df['attack'] = df['attack'].apply(lambda x: 0 if x == 'normal' else 1)
```

```python
df['attack'] = df['attack'].astype('str')
df['logged_in'] = df['logged_in'].astype('str')
df['is_host_login'] = df['is_host_login'].astype('str')
df['is_guest_login'] = df['is_guest_login'].astype('str')
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

train_features = NUMERIC_FEATURES + CAT_FEATURES
X = df[train_features]
y = df['attack']
```

```python
X.shape
```

```python
y.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, random_state=32)
```

```python
x_train.columns
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
x_test.shape
```

```python
import torch

from pytorch_tabular import TabularModel
from pytorch_tabular.models.ft_transformer.config import FTTransformerConfig
from pytorch_tabular.config import DataConfig, OptimizerConfig, TrainerConfig
```

```python
import torch
print(f"CUDA Available: {torch.cuda.is_available()}")
print(f"GPU Name: {torch.cuda.get_device_name(0)}")
```

```python
data_config = DataConfig(
    target=["attack"],
    continuous_cols=NUMERIC_FEATURES,  # Numerical features
    categorical_cols=CAT_FEATURES,  # Categorical features
)

trainer_config = TrainerConfig(
    accelerator="gpu",
    devices=-1,
    precision=16,
    auto_lr_find=False,
    batch_size=32,
    max_epochs=4,
)

optimizer_config = OptimizerConfig()

model_config = FTTransformerConfig(
    task="classification",
    input_embed_dim=32,  # Embedding dimension
    num_heads=4,  # Number of attention heads
    num_attn_blocks=6,  # Number of transformer blocks
    learning_rate=5e-5,
)


tabular_model = TabularModel(
    data_config=data_config,
    model_config=model_config,
    optimizer_config=optimizer_config,
    trainer_config=trainer_config,
)

# trainer = x_train.join(y_train)
trainer = pd.concat([x_train, y_train.rename("attack")], axis=1)
# trainer.shape
tabular_model.fit(train=trainer)
```

```python
preds = tabular_model.predict(x_test)
```

```python
preds['attack_prediction'].dtype
```

```python
preds.head()
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, preds['attack_prediction'])
print(f"Accuracy: {accuracy:.4f}")
print(f"Accuracy percentage (x100): {accuracy * 100:.4f}")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Assuming `y_test` contains the true labels for `x_test`
y_true = y_test.values.ravel()  # Convert to 1D array if needed
y_pred = preds["attack_prediction"].values  # Extract predicted labels

# Compute performance metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average="weighted")
recall = recall_score(y_true, y_pred, average="weighted")
f1 = f1_score(y_true, y_pred, average="weighted")

# Print metrics
print(f"Accuracy: {accuracy:.6f}")
print(f"Precision: {precision:.6f}")
print(f"Recall: {recall:.6f}")
print(f"F1 Score: {f1:.6f}")

# Display classification report
print("\nClassification Report:\n", classification_report(y_true, y_pred))

# Display confusion matrix
print("\nConfusion Matrix:\n", confusion_matrix(y_true, y_pred))
```
